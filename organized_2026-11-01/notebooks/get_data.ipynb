```python
import numpy as np

# Function to convert a protein sequence to one-hot encoding
def protein_to_one_hot(protein_sequence):
    # Define the mapping of amino acids to one-hot encoding
    amino_acids = "ABCDEFGHIJKLMNOPQRST"  # The 20 foldseek amino acids
    one_hot_mapping = {aa: [1 if aa == amino else 0 for amino in amino_acids] for aa in amino_acids}

    # Convert the protein sequence to one-hot encoding
    one_hot_sequence = np.array([one_hot_mapping[aa] for aa in protein_sequence])

    return one_hot_sequence

# Example usage
protein_sequence = "AAACCC"
one_hot_sequence = protein_to_one_hot(protein_sequence)
print(one_hot_sequence)
```
