```python
import requests

def retrieve_go_annotations(uniprot_id):
    base_url = f"https://www.ebi.ac.uk/proteins/api/proteins/{uniprot_id}"
    params = {
        'fields': 'go'
    }

    response = requests.get(base_url, params=params)

    if response.status_code == 200:
        data = response.json()
        go_annotations = data.get('go', [])
        return go_annotations
    else:
        print(f"Error: {response.status_code}")
        return None

# 要检索的 UniProt ID
your_uniprot_id = 'P12345'  # 替换为你的 UniProt ID

# 检索 GO 注释信息
go_info = retrieve_go_annotations(your_uniprot_id)

# 打印 GO 注释信息
if go_info:
    print(f"GO Annotations for UniProt ID {your_uniprot_id}:")
    for annotation in go_info:
        print(annotation)
```
