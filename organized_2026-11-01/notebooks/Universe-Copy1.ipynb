```python
def read_fasta(file_path):
    sequences = {}
    current_id = None
    with open(file_path, 'r') as file:
        for line in file:
            line = line.strip()
            if line.startswith('>'):
                current_id = line[1:]
                sequences[current_id] = ''
            else:
                sequences[current_id] = line
    return sequences

def remove_duplicates(input_f, output_file):
    sequences = read_fasta(input_f)
    unique_sequences = {}
    for sequence_id, sequence in sequences.items():
        if sequence not in unique_sequences.values():
            if sequence_id.split(' ')[-1] == 'Flag=1' or sequence_id.split(' ')[-1] == 'Flag=2':
                unique_sequences[sequence_id.split(' ')[0]] = sequence
    
    
    with open(output_file, 'a') as file:
        for sequence_id, sequence in unique_sequences.items():
            file.write('>' + sequence_id + '\n')
            file.write(sequence + '\n')

dirs = '../local_snapshot/data/'            


output_file = dirs + 'Universe_1000_2000_all.fasta'
i = 0
for input_file in Uni_list:
    inputf = dirs + 'Universe_1000_2000/' + input_file
    print(input_file)
    remove_duplicates(inputf, output_file)
    print("Duplicates removed from", i, "and appended to", output_file)
    i += 1
```

```python
import os
import numpy as np
def get_pLDDT(pdb_p):
    infile = open(pdb_p,'r')
    pLDDT = []
    for line in infile:
          if line[0:6].rstrip() == 'ATOM' and line[13:15].rstrip() == 'CA':
            pLDDT.append(float(line[60:65].rstrip()))
    infile.close()
    return np.mean(pLDDT)
af_path = '../local_snapshot/data/afall/'
af_list = os.listdir(af_path)
rep_structure = []
rep_pLDDT = []
keys = []
for key in rep_seq.keys():
    pLDDTs = []
    items = []
    for item in rep_seq[key]:
        item_name = item + '.pdb'
        if item_name in af_list:
            pdb_p = af_path+ item + '.pdb'
            pLDDTs.append(get_pLDDT(pdb_p))
            items.append(item)
    if len(pLDDTs)>0:
        max_plddt = max(pLDDTs)
        rep_pLDDT.append(max_plddt)
        rep_structure.append(items[pLDDTs.index(max_plddt)])
        keys.append(key)
```
